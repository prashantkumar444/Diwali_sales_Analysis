```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel("Diwali_sales_excel.xlsx")
```

```python
print(df.head())
print(df.tail())
print(df.shape)
```

```python
print(df.columns)
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
df.isnull().sum()
```

```python
#1-Sales by Gender
print(df.groupby('Gender')['Amount'].sum())
Gender_sales=df.groupby('Gender')['Amount'].sum()
plt.figure(figsize=(3,2))
Gender_sales.plot(kind='bar')
plt.title("Total sales by gender")
plt.xlabel('Gender')
plt.ylabel('Sales Amount')
plt.show()
```

```python
#2-Sales by Age
age_sales=df.groupby('Age_group')['Amount'].sum().sort_values(ascending=False)
print(age_sales)
plt.figure(figsize=(4,3))
age_sales.plot(kind='bar')
plt.title('Total sales by age groups')
plt.xlabel('Age Groups')
plt.ylabel('Sales Amount')
plt.show()
```

```python
#3-Sales by states
state_sale=df.groupby('State')['Amount'].sum().sort_values(ascending=False)
print(state_sale)
plt.figure(figsize=(4,3))
state_sale.plot(kind='bar')
plt.title('Sales by states')
plt.xlabel('States')
plt.ylabel('Sales Amount')
#plt.xticks(rotation=45)--used for rotating states by 45 degree
plt.show()
```

```python
#4-Sales by occupation
Occupation_sales=df.groupby('Occupation')['Amount'].sum().sort_values(ascending=False)
print(Occupation_sales)
plt.figure(figsize=(4,3))
Occupation_sales.plot(kind='bar')
plt.title('Sales by Occupation')
plt.xlabel('Occupation')
plt.ylabel('Sales Amount')
plt.show()
```

```python
#5- Top 5 selling products
Top_5_products=df.groupby('Product_name')['Amount'].sum().sort_values(ascending=False).head(5)
print(Top_5_products)
```

```python
#6-which payment mode is used most by the customers 
print(df['Payment_mode'].unique())
payment_count=df['Payment_mode'].value_counts()
print(payment_count)
#Hence UPI is mostly used by the customers 
```

```python
#7-Find the percentage of each payment mode
payment_percentage=df['Payment_mode'].value_counts(normalize=True)*100
print(payment_percentage)
#Conclution
#Here UPI is the preferred payment medium,accounting  for 32% of all transactions.
#The company should ensure a smooth UPI payment experience and may introduce UPI cashback offers to increase customer satisfactions.
payment_percentage=df['Payment_mode'].value_counts(normalize=True)*100
payment_percentage.plot(kind='pie',autopct='%1.1f%%',figsize=(5,5))
plt.title('Payment Mode Distribution')
plt.ylabel(' ')
plt.show()
```

```python
#8-Rating Analysis
print(df['Rating'].describe())
print(df['Rating'].value_counts()) # How many times did each rating occur?
print(df['Rating'].mean()) #what is the average rating?
rating=df['Rating'].value_counts().sort_index()
plt.figure(figsize=(4,3))
rating.plot(kind='bar')# Bar chart plotting
plt.title('Rating Distribution')
plt.xlabel('Rating')
plt.ylabel('Number of customers')
plt.show()

rating.plot(kind='pie',autopct='%1.1f%%')# Pie chart plotting
plt.title('Rating Distribution ')
plt.ylabel(' ')
plt.show()

#conclution 
# 88% of customers gave 4 or 5 star rating.
# This indicates that overall customer satisfaction is high.
#The business should identify why some customers gave 4-star and improve those areas to increase  5-Star rating.
```

```python
#9-Correlation analysis
print(df.corr(numeric_only=True)) #Find the correlation between numeric columns.
corr=df.corr(numeric_only=True)#Visualize the correlation matrix.

plt.figure(figsize=(6,5))
plt.imshow(corr,cmap='coolwarm')
plt.colorbar()

plt.xticks(range(len(corr.columns)),corr.columns,rotation=90)
plt.yticks(range(len(corr.columns)),corr.columns)
plt.title('Correlation Matrix')
plt.show()
```

```python
# We can also use Seaborn heatmap for correlation analysis(Give more cleaner and professional look)
corr=df.corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.heatmap(corr,annot=True,# Show correlation values
            cmap='coolwarm',# color theme
            linewidths=0.5)
plt.title('Correlation matrix')
plt.show()
#Conclution
#Price and Amount show the strongest positive correlation,indicating that higher-priced products contribute to higher sales amount.
#Quantity has a negative corelation with Price and Amount,suggesting that customers buying more items tend to purchase lower priced products.
#Age has little or no influence on purchasing behavior in this dataset.
#Rating have a slight positive relatioship with Price and Discount,indicating that customers may be more satisfied with higher-value purchases or attractive discounts.
#User_id has no meaningful business relationship and is not useful for analysis.
```
